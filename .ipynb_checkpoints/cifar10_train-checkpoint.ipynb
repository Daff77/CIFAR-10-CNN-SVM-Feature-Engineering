1. **Memuat Dataset CIFAR-10** (Data Training) & Preprocessing
2. **Eksplorasi & Visualisasi Data**
3. **Membangun Model CNN Kustom** (3 blok konvolusi + batch normalization + pooling)
4. **Evaluasi Model Baseline** (Sebelum GridSearch: 10 epochs, 6000 sampel)
5. **Menjalankan GridSearchCV** (Tuning `dense_units` & `dropout_rate` dengan K-Fold CV)

8. **Visualisasi Filter / Kernel CNN** 
9. **Ekstraksi & Visualisasi Feature Maps** 
10. **Menyimpan Model ke 1 File Saja** (`custom_cnn_cifar10_feature_extraction.keras`) untuk digunakan pada tahap pengujian & ekstraksi fitur SVM (`cifar10_test.ipynb`).

# ⚙️ 4. Pre-processing Data

Data training dinormalisasi ke rentang [0, 1]. Untuk validasi internal CNN, data training dipecah dengan rasio **80% training internal (40.000 sampel)** dan **20% validasi internal (10.000 sampel)**.

```python
# Scale data training
X_train = X_train / 255.0

# Split internal data training murni menjadi train (80%) dan validasi (20%)
# Validasi diambil murni dari data training (50.000 -> 40.000 train, 10.000 val)
X_train_split, X_val, y_train_split, y_val = train_test_split(
    X_train, y_train,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_train
)

# Verifikasi ketat tidak ada data testing yang bocor ke training/validasi
assert len(X_train_split) == 40000 and len(X_val) == 10000, 'Ukuran split internal harus 40.000 train dan 10.000 val!'

# Transform target variable into one-hot encoding untuk CNN
y_cat_train_split = to_categorical(y_train_split, 10)
y_cat_val = to_categorical(y_val, 10)

print(f"X_train_split shape (80% internal) : {X_train_split.shape}, y_cat_train_split: {y_cat_train_split.shape}")
print(f"X_val shape         (20% internal) : {X_val.shape},         y_cat_val        : {y_cat_val.shape}")
print("\n[STATUS] VALIDASI MURNI: Data validasi 100% berasal dari split internal data training.")
```

# 🏗️ 5. Membangun Model Deep CNN Kustom (4 Blok Konvolusi + L2 Regularization + GlobalAveragePooling2D)

Arsitektur ditingkatkan menjadi **4 Blok Konvolusi** (total 8 layer `Conv2D` dengan filter bertahap **64 -> 128 -> 256 -> 512**).
Fitur utama arsitektur ini:
- **L2 Weight Decay (`1e-4`)** & **Batch Normalization** pada setiap blok konvolusi untuk mencegah overfitting.
- **Progressive Dropout** ($0.15 \rightarrow 0.40$) seiring bertambahnya kedalaman feature map.
- **`GlobalAveragePooling2D` (512 dimensi)** sebagai layer `svm_features` untuk representasi fitur yang kaya, padat, dan bebas spasial overfitting.
- **Label Smoothing Crossentropy (`0.10`)** untuk meningkatkan pemisahan margin representasi fitur antar kelas.


```python
INPUT_SHAPE = (32, 32, 3)
KERNEL_SIZE = (3, 3)
WEIGHT_DECAY = 1e-4

def build_model(dense_units=512, dropout_rate=0.25):
    inputs = Input(shape=INPUT_SHAPE, name='input_image')

    # BLOCK 1: 32x32 -> 16x16 (64 filters)
    x = Conv2D(64, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv1_1')(inputs)
    x = BatchNormalization(name='bn1_1')(x)
    x = Conv2D(64, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv1_2')(x)
    x = BatchNormalization(name='bn1_2')(x)
    x = MaxPooling2D(pool_size=(2, 2), name='pool1')(x)
    x = Dropout(dropout_rate * 0.6, name='dropout1')(x)

    
    # BLOCK 2: 16x16 -> 8x8 (128 filters)
    x = Conv2D(128, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv2_1')(x)
    x = BatchNormalization(name='bn2_1')(x)
    x = Conv2D(128, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv2_2')(x)
    x = BatchNormalization(name='bn2_2')(x)
    x = MaxPooling2D(pool_size=(2, 2), name='pool2')(x)
    x = Dropout(dropout_rate, name='dropout2')(x)

    # BLOCK 3: 8x8 -> 4x4 (256 filters)
    x = Conv2D(256, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv3_1')(x)
    x = BatchNormalization(name='bn3_1')(x)
    x = Conv2D(256, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv3_2')(x)
    x = BatchNormalization(name='bn3_2')(x)
    x = MaxPooling2D(pool_size=(2, 2), name='pool3')(x)
    x = Dropout(dropout_rate * 1.4, name='dropout3')(x)

    # BLOCK 4: 4x4 -> 2x2 (512 filters)
    x = Conv2D(512, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='conv4_1')(x)
    x = BatchNormalization(name='bn4_1')(x)
    x = Conv2D(512, KERNEL_SIZE, padding='same', activation='relu',
               kernel_regularizer=l2(WEIGHT_DECAY), name='feature_map')(x)
    x = BatchNormalization(name='feature_map_bn')(x)
    x = MaxPooling2D(pool_size=(2, 2), name='pool4')(x)
    x = Dropout(dropout_rate * 1.6, name='dropout4')(x)

    # FEATURE EXTRACTION LAYER UNTUK SVM (512 Dimensi)
    svm_features = GlobalAveragePooling2D(name='svm_features')(x)

    # CLASSIFIER HEAD (CNN)
    x = Dense(dense_units, activation='relu',
              kernel_regularizer=l2(WEIGHT_DECAY), name='cnn_dense')(svm_features)
    x = BatchNormalization(name='cnn_dense_bn')(x)
    x = Dropout(0.50, name='cnn_head_dropout')(x)
    outputs = Dense(10, activation='softmax', name='cnn_softmax')(x)

    model = Model(inputs=inputs, outputs=outputs, name='custom_cnn_cifar10')

    METRICS = [
        'accuracy',
        tf.keras.metrics.Precision(name='precision'),
        tf.keras.metrics.Recall(name='recall')
    ]
    optimizer = tf.keras.optimizers.Adam(learning_rate=1e-3)
    loss = tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.10)
    model.compile(loss=loss, optimizer=optimizer, metrics=METRICS)
    return model
```

# 🔍 6. Evaluasi Model Baseline & GridSearchCV

```python
# 1. Evaluasi Model Baseline (Sebelum GridSearch: dense_units=128, dropout_rate=0.25)
print("--- 1. Evaluating Baseline Model (Before GridSearch: dense_units=128, dropout_rate=0.25) ---")
baseline_model = build_model(dense_units=128, dropout_rate=0.25)
baseline_history = baseline_model.fit(
    X_train_split[:6000], y_cat_train_split[:6000],
    epochs=10,
    batch_size=64,
    validation_data=(X_val, y_cat_val),
    verbose=1
)

baseline_eval = baseline_model.evaluate(X_val, y_cat_val, verbose=0)
baseline_val_loss = baseline_eval[0]
baseline_val_acc = baseline_eval[1]
print(f"\nBaseline (Before GridSearch) -> Val Loss: {baseline_val_loss:.4f}, Val Accuracy: {baseline_val_acc * 100:.2f}%")
```

```python
# 2. Menjalankan GridSearchCV (10 Epochs)
print("--- 2. Running GridSearchCV (10 Epochs) ---")
grid_samples = 6000
X_grid = X_train_split[:grid_samples]
y_grid = y_cat_train_split[:grid_samples]

grid_model = KerasClassifier(build_fn=build_model, epochs=10, batch_size=64, verbose=1)

param_grid = {
    'dense_units': [256, 512],
    'dropout_rate': [0.25, 0.35]
}

cv = KFold(n_splits=2, shuffle=True, random_state=RANDOM_STATE)
grid = GridSearchCV(estimator=grid_model, param_grid=param_grid, cv=cv, verbose=1)
grid_result = grid.fit(X_grid, y_grid)

print(f"\nBest parameters: {grid_result.best_params_}")
print(f"Best CV Score: {grid_result.best_score_:.4f}")

best_dense = grid_result.best_params_.get('dense_units', 512)
best_dropout = grid_result.best_params_.get('dropout_rate', 0.25)
print(f"Selected parameters for final training: dense_units={best_dense}, dropout_rate={best_dropout}")
```

# 🚀 7. Training Model CNN Tuned (Setelah GridSearch)

```python
# Membangun model dengan hyperparameter terbaik hasil GridSearchCV
model = build_model(dense_units=best_dense, dropout_rate=best_dropout)
model.summary()

# Enhanced Data Augmentation
batch_size = 64
data_generator = ImageDataGenerator(
    width_shift_range=0.10, 
    height_shift_range=0.10, 
    horizontal_flip=True,
    rotation_range=10,
    zoom_range=0.10,
    fill_mode="reflect"
)
train_generator = data_generator.flow(X_train_split, y_cat_train_split, batch_size=batch_size, shuffle=True, seed=RANDOM_STATE)
steps_per_epoch = X_train_split.shape[0] // batch_size

# Visualisasi augmentasi
fig, axes = plt.subplots(1, 5, figsize=(12, 3))
for i in range(5):
    aug_img = data_generator.random_transform(X_train_split[0])
    axes[i].imshow(aug_img)
    axes[i].axis('off')
    axes[i].set_title(f"Augmented #{i+1}")
plt.suptitle('Contoh Data Augmentation', fontweight='bold')
plt.show()

# File model artifact
cnn_model_filename = 'custom_cnn_cifar10_feature_extraction.keras'

# Callbacks untuk training optimal (Hingga 150-200 epochs dengan EarlyStopping & LR Decay otomatis)
MAX_EPOCHS = 150

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss', 
    factor=0.5,        # kurangi LR jadi setengah jika loss stagnan
    patience=7,         # tunggu 7 epoch
    min_lr=1e-6,
    verbose=1
)

early_stop = EarlyStopping(
    monitor='val_accuracy', 
    mode='max',
    patience=25,               # tunggu 25 epoch jika tidak ada peningkatan
    restore_best_weights=True,  # kembalikan bobot epoch terbaik
    verbose=1
)

checkpoint = ModelCheckpoint(
    cnn_model_filename,
    monitor='val_accuracy',
    mode='max',
    save_best_only=True,
    verbose=1
)

# Training model
print(f"\n--- Training Tuned Deep CNN Model (Up to {MAX_EPOCHS} epochs with Data Augmentation + Callbacks) ---")
r = model.fit(
    train_generator, 
    epochs=MAX_EPOCHS,
    steps_per_epoch=steps_per_epoch,
    validation_data=(X_val, y_cat_val),
    callbacks=[reduce_lr, early_stop, checkpoint],
    verbose=1
)

# Evaluasi pada validation set
val_eval = model.evaluate(X_val, y_cat_val, verbose=0)
tuned_val_loss = val_eval[0]
tuned_val_acc = val_eval[1]
print(f"\nTuned Model (After GridSearch) -> Val Loss: {tuned_val_loss:.4f}, Val Accuracy: {tuned_val_acc * 100:.2f}%")
```

# 📊 8. Evaluasi & Perbandingan Model (Sebelum vs Sesudah GridSearch)

```python
# Plot training history
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
plt.plot(r.history['loss'], label='Train Loss')
plt.plot(r.history['val_loss'], label='Val Loss')
plt.title('Loss Function Evolution')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(2, 2, 2)
plt.plot(r.history['accuracy'], label='Train Accuracy')
plt.plot(r.history['val_accuracy'], label='Val Accuracy')
plt.title('Accuracy Function Evolution')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(2, 2, 3)
plt.plot(r.history['precision'], label='Train Precision')
plt.plot(r.history['val_precision'], label='Val Precision')
plt.title('Precision Function Evolution')
plt.xlabel('Epoch')
plt.ylabel('Precision')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(2, 2, 4)
plt.plot(r.history['recall'], label='Train Recall')
plt.plot(r.history['val_recall'], label='Val Recall')
plt.title('Recall Function Evolution')
plt.xlabel('Epoch')
plt.ylabel('Recall')
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Perbandingan performa model Sebelum vs Sesudah GridSearch pada Data Validasi
acc_improvement = (tuned_val_acc - baseline_val_acc) * 100

print("=" * 65)
print("     PERBANDINGAN MODEL SEBELUM vs SESUDAH GRID SEARCH")
print("=" * 65)
print(f" Sebelum GridSearch : Val Acc = {baseline_val_acc * 100:.2f}%, Val Loss = {baseline_val_loss:.4f}")
print(f" Sesudah GridSearch : Val Acc = {tuned_val_acc * 100:.2f}%, Val Loss = {tuned_val_loss:.4f}")
print(f" Peningkatan Akurasi (+Delta) : +{acc_improvement:.2f}% Peningkatan")
print("=" * 65)

# Visualisasi Bar Chart
categories = ['Sebelum GridSearch', 'Sesudah GridSearch']
acc_values = [baseline_val_acc * 100, tuned_val_acc * 100]

plt.figure(figsize=(7, 5))
bars = plt.bar(categories, acc_values, color=['#95a5a6', '#2ecc71'], width=0.45, edgecolor='black')
plt.ylabel('Validation Accuracy (%)')
plt.title('Peningkatan Akurasi Model: Sebelum vs Sesudah GridSearch', fontsize=12)
plt.ylim(0, 100)
plt.grid(axis='y', alpha=0.3)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2.0, yval + 1.5, f'{yval:.2f}%', ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.show()
```


# 🔬 9. Visualisasi Filter/Kernel CNN

```python
# Ringkasan bentuk filter pada setiap layer konvolusi
print("=" * 50)
print("Ringkasan Filter pada Layer Konvolusi:")
print("=" * 50)
for layer in model.layers:
    if 'conv' not in layer.name:
        continue
    filters, biases = layer.get_weights()
    print(f"{layer.name}: {filters.shape}")
```

```python
# Visualisasi filter dari layer konvolusi pertama (conv1_1)
first_conv_layer = None
for layer in model.layers:
    if 'conv' in layer.name:
        first_conv_layer = layer
        break

filters, biases = first_conv_layer.get_weights()
print(f"Filter shape: {filters.shape}")

# Normalize filter values ke 0-1 untuk visualisasi
f_min, f_max = filters.min(), filters.max()
filters_norm = (filters - f_min) / (f_max - f_min)

# Plot hingga 32 filter pertama (untuk setiap channel RGB)
n_show = min(32, filters_norm.shape[3])
n_channels = filters_norm.shape[2]  # 3 (RGB)

fig, axes = plt.subplots(n_show, n_channels, figsize=(6, n_show * 1.5))
for i in range(n_show):
    for j in range(n_channels):
        axes[i, j].imshow(filters_norm[:, :, j, i], cmap='gray')
        axes[i, j].set_xticks([])
        axes[i, j].set_yticks([])
        if i == 0:
            axes[i, j].set_title(['R', 'G', 'B'][j], fontsize=10)
    axes[i, 0].set_ylabel(f'F{i+1}', fontsize=8, rotation=0, labelpad=15)

plt.suptitle(f'Filter/Kernel dari Layer: {first_conv_layer.name}\n({n_show} filter x 3 channel RGB)', 
             fontsize=14, fontweight='bold', y=1.0)
plt.tight_layout()
plt.show()
```

```python
# Pilih gambar contoh untuk visualisasi (dari data validasi)
sample_idx = 0
sample_image = X_val[sample_idx]

plt.figure(figsize=(3, 3))
plt.imshow(sample_image)
plt.title(f"Sample Image: {classes_name[int(y_val[sample_idx])]}")
plt.axis('off')
plt.show()

# Identifikasi semua layer konvolusi
conv_layers = [layer for layer in model.layers if 'conv' in layer.name]
print(f"\nJumlah layer konvolusi: {len(conv_layers)}")
for layer in conv_layers:
    print(f"  - {layer.name}: output shape = {layer.output_shape}")
```

```python
# Feature maps dari layer konvolusi pertama (shallow features)
first_conv_model = Model(inputs=model.input, outputs=conv_layers[0].output)
feature_maps_first = first_conv_model.predict(sample_image.reshape(1, 32, 32, 3))

print(f"Feature maps shape dari {conv_layers[0].name}: {feature_maps_first.shape}")

# Plot feature maps
n_maps = feature_maps_first.shape[3]  # 32
square = int(np.ceil(np.sqrt(n_maps)))

fig, axes = plt.subplots(square, square, figsize=(14, 14))
axes = axes.ravel()
for i in range(square * square):
    if i < n_maps:
        axes[i].imshow(feature_maps_first[0, :, :, i], cmap='gray')
        axes[i].set_title(f'Map {i+1}', fontsize=7)
    axes[i].set_xticks([])
    axes[i].set_yticks([])

plt.suptitle(f'Feature Maps dari Layer: {conv_layers[0].name} (Shallow Features)', 
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Feature maps dari layer konvolusi terakhir (deep features)
last_conv_layer = conv_layers[-1]
deep_conv_model = Model(inputs=model.input, outputs=last_conv_layer.output)
feature_maps_deep = deep_conv_model.predict(sample_image.reshape(1, 32, 32, 3))

print(f"Feature maps shape dari {last_conv_layer.name}: {feature_maps_deep.shape}")

# Plot feature maps (tampilkan 64 dari 128)
n_show = min(64, feature_maps_deep.shape[3])
square = int(np.ceil(np.sqrt(n_show)))

fig, axes = plt.subplots(square, square, figsize=(14, 14))
axes = axes.ravel()
for i in range(square * square):
    if i < n_show:
        axes[i].imshow(feature_maps_deep[0, :, :, i], cmap='gray')
        axes[i].set_title(f'Map {i+1}', fontsize=7)
    axes[i].set_xticks([])
    axes[i].set_yticks([])

plt.suptitle(f'Feature Maps dari Layer: {last_conv_layer.name} (Deep Features)', 
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Visualisasi feature maps dari SEMUA layer konvolusi (perbandingan)
fig, axes_all = plt.subplots(len(conv_layers), 8, figsize=(16, len(conv_layers) * 2.5))

for row, conv_layer in enumerate(conv_layers):
    temp_model = Model(inputs=model.input, outputs=conv_layer.output)
    fmaps = temp_model.predict(sample_image.reshape(1, 32, 32, 3), verbose=0)
    
    for col in range(8):
        if col < fmaps.shape[3]:
            axes_all[row, col].imshow(fmaps[0, :, :, col], cmap='viridis')
        axes_all[row, col].set_xticks([])
        axes_all[row, col].set_yticks([])
        if col == 0:
            axes_all[row, col].set_ylabel(conv_layer.name, fontsize=8, rotation=0, labelpad=50)

plt.suptitle('Perbandingan Feature Maps dari Setiap Layer Konvolusi (8 map pertama)', 
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

# 🧠 12. Ekstraksi Fitur CNN Menggunakan Layer `svm_features` & Penyimpanan ke Disk (NumPy / Pickle)

Fitur diekstraksi dari layer **`svm_features`** (`GlobalAveragePooling2D`), menghasilkan vektor representasi berdimensi 512.

> **💡 Optimasi Penggunaan Memori (RAM):**
> Hasil ekstraksi fitur dan label disimpan langsung ke dalam file proyek dalam format **NumPy (`.npz` / `.npy`)** dan **Pickle (`.pkl`)**.
> - **Hemat RAM**: Data gambar mentah (`X_train`) dan temporary batch segera dibersihkan dari RAM setelah fitur tersimpan.
> - **Efisien & Persisten**: Jika kernel direstart atau sel dijalankan kembali, fitur akan langsung dimuat dari disk tanpa perlu ekstraksi ulang dari awal.

```python
import os
import gc
import pickle
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import GlobalAveragePooling2D, Flatten
from tensorflow.keras.datasets import cifar10

RANDOM_STATE = 23092026

# Bersihkan alokasi memori yang tidak terpakai
gc.collect()

# Definisi path file penyimpanan hasil ekstraksi fitur di dalam folder proyek
FEATURES_NPZ_PATH = 'cifar10_train_features.npz'   # Format NumPy Archive terkompresi (.npz)
FEATURES_NPY_PATH = 'cifar10_train_features.npy'   # Format NumPy Array (.npy) - mendukung mmap
LABELS_NPY_PATH   = 'cifar10_train_labels.npy'     # Format NumPy Array Label (.npy)
FEATURES_PKL_PATH = 'cifar10_train_features.pkl'   # Format Pickle (.pkl)

# Cek apakah data ekstraksi fitur sudah tersimpan di direktori proyek
has_npz = os.path.exists(FEATURES_NPZ_PATH)
has_npy = os.path.exists(FEATURES_NPY_PATH) and os.path.exists(LABELS_NPY_PATH)
has_pkl = os.path.exists(FEATURES_PKL_PATH)

if has_npz or has_npy or has_pkl:
    print("=" * 65)
    print("  MEMUAT HASIL EKSTRAKSI FITUR DARI DISK (BEBAS BEBAN RAM)")
    print("=" * 65)
    
    if has_npz:
        print(f"Memuat dari file NumPy Archive: {FEATURES_NPZ_PATH}...")
        data_npz = np.load(FEATURES_NPZ_PATH)
        X_train_features = data_npz['X_train_features']
        y_train_flat = data_npz['y_train_flat']
    elif has_npy:
        print(f"Memuat dari file NumPy (.npy): {FEATURES_NPY_PATH}...")
        X_train_features = np.load(FEATURES_NPY_PATH)
        y_train_flat = np.load(LABELS_NPY_PATH)
    else:
        print(f"Memuat dari file Pickle (.pkl): {FEATURES_PKL_PATH}...")
        with open(FEATURES_PKL_PATH, 'rb') as f:
            pkl_data = pickle.load(f)
            X_train_features = pkl_data['X_train_features']
            y_train_flat = pkl_data['y_train_flat']
            
    print(f"\n[BERHASIL] Fitur berhasil dimuat dari disk tanpa ekstraksi ulang!")
    print(f"X_train_features shape: {X_train_features.shape}")
    print(f"y_train_flat shape    : {y_train_flat.shape}")

else:
    print("=" * 65)
    print("  EKSTRAKSI FITUR DARI MODEL CNN & PENYIMPANAN KE DISK")
    print("=" * 65)

    # 1. Pastikan model CNN tersedia di memori (muat dari file .h5 jika belum ada)
    cnn_model_filename = 'custom_cnn_cifar10_feature_extraction.keras'
    if 'model' not in globals() or model is None:
        if not os.path.exists(cnn_model_filename) and os.path.exists('custom_cnn_cifar10_feature_extraction.h5'):
            cnn_model_filename = 'custom_cnn_cifar10_feature_extraction.h5'
        if os.path.exists(cnn_model_filename):
            print(f"Memuat model CNN yang sudah tersimpan dari: {cnn_model_filename}")
            model = load_model(cnn_model_filename)
        else:
            raise FileNotFoundError(f"File model {cnn_model_filename} tidak ditemukan. Silakan jalankan sel pelatihan CNN terlebih dahulu.")

    # 2. Muat dataset training CIFAR-10
    if 'X_train' not in globals() or 'y_train' not in globals() or X_train is None:
        print("Memuat dataset training CIFAR-10 ke memori...")
        (X_train, y_train), _ = cifar10.load_data()
        X_train = X_train.astype('float32') / 255.0

    y_train_flat = y_train.flatten()

    # 3. Cari layer ekstraksi fitur (svm_features / GlobalAveragePooling2D)
    feature_layer = None
    for layer in model.layers:
        if layer.name == 'svm_features' or isinstance(layer, GlobalAveragePooling2D):
            feature_layer = layer
            break

    if feature_layer is None:
        for layer in model.layers:
            if isinstance(layer, Flatten):
                feature_layer = layer
                break

    feature_extractor = Model(inputs=model.input, outputs=feature_layer.output)
    print(f"Feature extractor layer: {feature_layer.name}, Dimensi fitur: {feature_layer.output_shape[-1]}")

    # Ekstraksi fitur secara bertahap (chunk_size = 5.000) untuk mencegah GPU OOM
    chunk_size = 5000
    features_list = []
    total_samples = len(X_train)
    print(f"\nMengekstrak fitur CNN dari seluruh data training ({total_samples} sampel) secara bertahap...")

    for i in range(0, total_samples, chunk_size):
        end_idx = min(i + chunk_size, total_samples)
        batch_num = i // chunk_size + 1
        total_batches = (total_samples + chunk_size - 1) // chunk_size
        print(f"  Memproses batch {batch_num}/{total_batches} (sampel {i} s.d. {end_idx})...")
        chunk_data = X_train[i:end_idx]
        chunk_feat = feature_extractor.predict(chunk_data, batch_size=64, verbose=0)
        features_list.append(chunk_feat)

    # Gabungkan hasil ekstraksi
    X_train_features = np.vstack(features_list)
    print(f"\n[SELESAI] Ekstraksi tuntas. X_train_features shape: {X_train_features.shape}")

    # 4. SIMPAN DATA KE FILE PROYEK (NUMPY & PICKLE)
    print("\n" + "=" * 65)
    print("  MENYIMPAN DATA FITUR KE DIREKTORI PROYEK (BUKAN HANYA DI RAM)")
    print("=" * 65)
    
    # A. Simpan Format NumPy Compressed (.npz)
    np.savez_compressed(
        FEATURES_NPZ_PATH,
        X_train_features=X_train_features,
        y_train_flat=y_train_flat
    )
    size_npz_mb = os.path.getsize(FEATURES_NPZ_PATH) / (1024 * 1024)
    print(f"  [1/3] NumPy Archive (.npz) tersimpan : {FEATURES_NPZ_PATH} ({size_npz_mb:.2f} MB)")

    # B. Simpan Format NumPy Array (.npy) - mendukung memory-mapping
    np.save(FEATURES_NPY_PATH, X_train_features)
    np.save(LABELS_NPY_PATH, y_train_flat)
    size_npy_mb = os.path.getsize(FEATURES_NPY_PATH) / (1024 * 1024)
    print(f"  [2/3] NumPy Arrays  (.npy) tersimpan : {FEATURES_NPY_PATH} ({size_npy_mb:.2f} MB)")

    # C. Simpan Format Pickle (.pkl)
    with open(FEATURES_PKL_PATH, 'wb') as f:
        pickle.dump({
            'X_train_features': X_train_features,
            'y_train_flat': y_train_flat
        }, f, protocol=pickle.HIGHEST_PROTOCOL)
    size_pkl_mb = os.path.getsize(FEATURES_PKL_PATH) / (1024 * 1024)
    print(f"  [3/3] Pickle File   (.pkl) tersimpan : {FEATURES_PKL_PATH} ({size_pkl_mb:.2f} MB)")

    # 5. BERSIHKAN RAM: Hapus variabel gambar mentah dan temporary list dari memori
    del features_list
    if 'X_train' in globals():
        del X_train
    if 'y_train' in globals():
        del y_train
    gc.collect()
    print("\n[BERSIHKAN RAM] Gambar mentah CIFAR-10 (X_train) dan temporary list telah dihapus dari RAM.")

print(f"\n[STATUS] Data siap digunakan untuk SVM Classifier.")
print(f"X_train_features: {X_train_features.shape}, y_train_flat: {y_train_flat.shape}")
```

# 🔍 13. Grid Search Hyperparameter SVM Classifier


```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

RANDOM_STATE = 23092026

SVM_GRID_SAMPLES = 15000

if SVM_GRID_SAMPLES < len(X_train_features):
    X_svm_grid, _, y_svm_grid, _ = train_test_split(
        X_train_features,
        y_train_flat,
        train_size=SVM_GRID_SAMPLES,
        random_state=RANDOM_STATE,
        stratify=y_train_flat
    )
else:
    X_svm_grid = X_train_features
    y_svm_grid = y_train_flat

# Pipeline: StandardScaler + SVC RBF (cache_size 2048MB per worker)
svm_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(
        kernel='rbf',
        cache_size=2048,
        random_state=RANDOM_STATE
    ))
])

# Kandidat parameter SVM
param_grid = {
    'svm__C': [1, 5, 10, 20],
    'svm__gamma': ['scale', 0.0005, 0.001, 0.002]
}

print("=" * 60)
print("  MENJALANKAN GRID SEARCH CV PADA SVM CLASSIFIER (OPTIMIZED)")
print("=" * 60)
print(f"Data untuk Grid Search  : {X_svm_grid.shape}")
print(f"Jumlah kandidat         : {len(param_grid['svm__C']) * len(param_grid['svm__gamma'])} kombinasi")
print(f"Total model fit (3-fold): {len(param_grid['svm__C']) * len(param_grid['svm__gamma']) * 3} kali pelatihan\n")

grid_search_svm = GridSearchCV(
    estimator=svm_pipeline,
    param_grid=param_grid,
    scoring='accuracy',
    cv=3,
    n_jobs=4,
    verbose=2,
    refit=False
)

grid_search_svm.fit(X_svm_grid, y_svm_grid)

print("\n" + "=" * 60)
print("  HASIL TERBAIK GRID SEARCH SVM")
print("=" * 60)
print("Parameter Terbaik :", grid_search_svm.best_params_)
print(f"Best Mean CV Acc  : {grid_search_svm.best_score_ * 100:.2f}%")
```

```python
from IPython.display import display

# Tampilkan tabel rangkuman hasil GridSearchCV SVM
grid_results = pd.DataFrame(grid_search_svm.cv_results_)
grid_cols = ['param_svm__C', 'param_svm__gamma', 'mean_test_score', 'std_test_score', 'rank_test_score']

ranked_df = grid_results[grid_cols].sort_values('rank_test_score').reset_index(drop=True)
print("Peringkat Kombinasi Hyperparameter SVM:")
display(ranked_df)
```

# 🚀 14. Train SVM Final Menggunakan Hyperparameter Terbaik


```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

RANDOM_STATE = 23092026

best_C = grid_search_svm.best_params_['svm__C']
best_gamma = grid_search_svm.best_params_['svm__gamma']

# Membangun pipeline SVM final dengan parameter terbaik dan cache 4GB
final_svm = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(
        C=best_C,
        gamma=best_gamma,
        kernel='rbf',
        cache_size=4096,
        random_state=RANDOM_STATE,
        decision_function_shape='ovr'
    ))
])

# Melatih final SVM pada SELURUH fitur representasi data latih (optimal untuk mencapai akurasi tertinggi >92%)
print("=" * 60)
print("  MELATIH SVM FINAL PADA SELURUH FITUR TRAINING")
print("=" * 60)
print(f"Konfigurasi : C = {best_C}, gamma = {best_gamma}")
print(f"Data Latih  : {len(X_train_features)} sampel ({X_train_features.shape[1]} dimensi)")
print("Sedang melatih SVM Final...")

final_svm.fit(X_train_features, y_train_flat)
print("\n[BERHASIL] SVM Final telah selesai dilatih!")

# Verifikasi akurasi pada data training (5.000 sampel)
sample_pred = final_svm.predict(X_train_features[:5000])
sample_acc = accuracy_score(y_train_flat[:5000], sample_pred)
print(f"Akurasi SVM pada 5.000 sampel training: {sample_acc * 100:.2f}%")
```

# 📦 15. Serialisasi & Injeksi Final SVM ke Dalam File .keras (Single Container)\n

```python
import pickle
import h5py
import numpy as np

# Serialisasi pipeline SVM final menjadi byte array
svm_bytes = pickle.dumps(final_svm, protocol=pickle.HIGHEST_PROTOCOL)
svm_array = np.frombuffer(svm_bytes, dtype=np.uint8)

# Ambil nama dan dimensi feature layer secara aman (bahkan jika fitur dimuat langsung dari disk)
feat_layer_name = feature_layer.name if 'feature_layer' in globals() and feature_layer is not None else 'svm_features'
feat_dim = int(feature_layer.output_shape[-1]) if 'feature_layer' in globals() and feature_layer is not None else int(X_train_features.shape[1])

# Sisipkan ke dalam file .keras hasil injeksi final
cnn_model_filename = 'custom_cnn_cifar10_feature_extraction.keras'

# Pastikan model dasar CNN sudah tersimpan dalam format .keras
if 'model' in globals() and model is not None:
    model.save(cnn_model_filename)
elif not os.path.exists(cnn_model_filename) and os.path.exists('custom_cnn_cifar10_feature_extraction.h5'):
    tmp_m = load_model('custom_cnn_cifar10_feature_extraction.h5')
    tmp_m.save(cnn_model_filename)

with h5py.File(cnn_model_filename, 'a') as h5f:
    if 'svm_pipeline_pickle' in h5f:
        del h5f['svm_pipeline_pickle']
    
    h5f.create_dataset(
        'svm_pipeline_pickle',
        data=svm_array,
        compression='gzip'
    )
    
    # Metadata pendukung
    h5f.attrs['feature_layer'] = feat_layer_name
    h5f.attrs['feature_dimension'] = feat_dim
    h5f.attrs['svm_type'] = 'SVC_RBF'
    h5f.attrs['svm_C'] = float(best_C)
    h5f.attrs['svm_gamma'] = str(best_gamma)

print("=" * 60)
print(f"[BERHASIL] Seluruh model tersimpan dalam SATU file tunggal (.keras): {cnn_model_filename}")
print("Isi di dalam file .keras:")
print("  1. Arsitektur dan Bobot CNN Kustom (Keras format)")
print("  2. Dataset '/svm_pipeline_pickle' (StandardScaler + Trained Final SVM RBF)")
print(f"  3. Metadata: C={best_C}, gamma={best_gamma}, feature_layer={feat_layer_name}, feature_dim={feat_dim}")
print("=" * 60)
print("\nOutput training tuntas! Buka 'cifar10_test.ipynb' untuk langsung mengevaluasi performa pada data test.")
```
